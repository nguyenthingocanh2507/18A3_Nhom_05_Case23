TASK 04_ Merge & Dữ liệu

Đọc dữ liệu sạch: 

```python
import pandas as pd

facility = pd.read_csv("../data_clean/facility_info_clean.csv")
incident = pd.read_csv("../data_clean/incident_report_clean.csv")
maintenance = pd.read_csv("../data_clean/maintenance_log_clean.csv")
```

4.1 - Kết hợp facility_info với incident_report dựa trên facility_id.

```python
incident_facility = incident.merge(
    facility,
    on="facility_id",
    how="left",
    indicator=True
)

incident_facility.head()
```

4.2 - Kết hợp incident_report với maintenance_log dựa trên report_id.

```python
incident_maintenance = incident.merge(
    maintenance,
    on="report_id",
    how="left",
    indicator=True
)

incident_maintenance.head()
```

4.3 - Phát hiện các sự cố không có bản ghi bảo trì và lý giải nguyên nhân. 

```python
no_maintenance = incident_maintenance[
    incident_maintenance["_merge"] == "left_only"
]

no_maintenance[["report_id", "facility_id", "severity"]]
```

Nhận xét một số sự cố không có bản ghi bảo trì, có thể do:
  - Sự cố nhẹ chưa cần sửa chữa
  - Dữ liệu bảo trì chưa được ghi nhận
  - Lỗi nhập liệu hoặc thiếu đồng bộ


4.5 - Xuất file dữ liệu hoàn chỉnh phục vụ phân tích pivot

```python
full_data = (
    incident
    .merge(facility, on="facility_id", how="left")
    .merge(maintenance, on="report_id", how="left")
)

full_data.head()
full_data.to_csv("../data_clean/full_merged_data.csv", index=False)
```
