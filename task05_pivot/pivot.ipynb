TASK 05_ Pivot Tables + Stack/Unstack

Đọc file đã merge:

```python
import pandas as pd

full_data = pd.read_csv("../data_clean/full_merged_data.csv")
```

5.1 - Tạo bảng pivot thể hiện thời gian sửa chữa trung bình theo facility_type × building. 

```python
pivot_repair_time = pd.pivot_table(
    full_data,
    values="duration_hours",
    index="facility_type",
    columns="building",
    aggfunc="mean"
)

pivot_repair_time
```

5.2 - Tạo bảng pivot về số lượng sự cố theo building × severity. 

```python
pivot_incident_count = pd.pivot_table(
    full_data,
    values="report_id",
    index="building",
    columns="severity",
    aggfunc="count",
    fill_value=0
)

pivot_incident_count
```

5.3 - Thực hành chuyển đổi giữa dạng rộng và dạng dài bằng stack() và unstack().

```python
# Dạng RỘNG --> DÀI
stacked = pivot_repair_time.stack().reset_index()
stacked.columns = ["facility_type", "building", "avg_duration_hours"]

stacked

# Dạng DÀI --> RỘNG
unstacked = stacked.set_index(
    ["facility_type", "building"]
)["avg_duration_hours"].unstack()

unstacked
```

5.4. Phân tích kết quả Pivot

1. Thời gian sửa chữa trung bình theo facility_type × building
- Từ bảng pivot cho thấy, **facility_type = …** tại **building = …** có
  thời gian sửa chữa trung bình cao nhất.
- Điều này cho thấy các cơ sở thuộc loại này tại building trên có mức độ
  hư hỏng phức tạp hoặc quy trình bảo trì chưa hiệu quả.

2. Số lượng sự cố theo building × severity
- Một số building như **…** có số lượng sự cố mức **severity = High**
  cao hơn các khu vực khác.
- Điều này phản ánh mức độ rủi ro vận hành và nhu cầu kiểm tra định kỳ
  cao tại các building này.

3. Đánh giá loại cơ sở cần ưu tiên bảo trì
- Kết hợp hai bảng pivot, có thể nhận thấy **facility_type = …** vừa có
  **thời gian sửa chữa trung bình cao**, vừa xuất hiện nhiều sự cố nghiêm trọng.
- Do đó, đây là loại cơ sở **cần được ưu tiên bảo trì và nâng cấp** trong
  kế hoạch quản lý CSVC.

4. Nhận xét tổng quát
- Phân tích pivot giúp phát hiện các điểm nóng về bảo trì theo nhiều chiều:
  loại cơ sở – tòa nhà – mức độ sự cố.
- Kết quả là cơ sở quan trọng cho việc ra quyết định phân bổ nguồn lực bảo trì.
