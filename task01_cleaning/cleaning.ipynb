- Đọc ba file dữ liệu và khảo sát số lượng, cấu trúc và các vấn đề tồn tại. 

```python
import pandas as pd

# ========== ĐỌC DỮ LIỆU ==========
facility = pd.read_csv("data_raw/facility_info.csv")
incident = pd.read_csv("data_raw/incident_report.csv")
maint = pd.read_csv("data_raw/maintenance_log.csv")

# ========== KHẢO SÁT DỮ LIỆU ==========
def survey(df, name):
    print(f"\n--- {name} ---")
    print("Shape:", df.shape)
    print("\nInfo:")
    print(df.info())
    print("\nMissing values:")
    print(df.isnull().sum())
    print("\nSample:")
    print(df.head())

survey(facility, "FACILITY INFO")
survey(incident, "INCIDENT REPORT")
survey(maint, "MAINTENANCE LOG")
```

- Chuẩn hóa facility_id để loại bỏ khoảng trắng, ký tự thừa và sự không thống nhất. 

```python
# Chuẩn hóa facility_id
for df in [facility, incident]:
    df["facility_id"] = (
        df["facility_id"]
        .astype(str)
        .str.strip()
        .str.upper()
    )
```

- Chuẩn hóa facility_type và building nhằm đảm bảo đồng nhất định dạng

```python
# Chuẩn hóa facility_type & building
facility["facility_type"] = facility["facility_type"].str.strip().str.title()
facility["building"] = facility["building"].str.strip().str.upper()
```

- Làm sạch capacity và đưa về kiểu dữ liệu phù hợp cho phân tích

```python
# Làm sạch capacity
facility["capacity"] = pd.to_numeric(facility["capacity"], errors="coerce")
```

- Chuẩn hóa date và severity trong incident_report. 

```python
# Chuẩn hóa incident
incident["date"] = pd.to_datetime(incident["date"], errors="coerce")
incident["severity"] = incident["severity"].str.strip().str.lower()
```

- Chuẩn hóa maintenance_date, duration_hours và technician trong maintenance_log. 

```python
# Chuẩn hóa maintenance
maint["maintenance_date"] = pd.to_datetime(maint["maintenance_date"], errors="coerce")
maint["duration_hours"] = pd.to_numeric(maint["duration_hours"], errors="coerce")
maint["technician"] = maint["technician"].str.strip().str.title()
```

- Xuất các file sạch để dùng cho các nhiệm vụ sau. 

```python
facility.to_csv("data_clean/facility_info_clean.csv", index=False)
incident.to_csv("data_clean/incident_report_clean.csv", index=False)
maint.to_csv("data_clean/maintenance_log_clean.csv", index=False)
```
